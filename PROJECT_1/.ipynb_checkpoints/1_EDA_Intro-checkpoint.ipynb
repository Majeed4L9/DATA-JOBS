```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ast 

import warnings
warnings.filterwarnings("ignore")
```

```python
raw_data = pd.read_csv(r'C:\Users\MAJEED\Desktop\APPS\data_jobs.csv')
raw_data.head()
```

```python
df = raw_data.copy()
```

```python
df.dropna(subset = ['job_skills'], inplace = True)
```

```python
def clean_list(skill_clean):
    return ast.literal_eval(skill_clean)
```

```python
df['job_skills'] = df['job_skills'].apply(clean_list)
```

```python
df_DA_US = df[(df['job_title_short'] == 'Data Analyst') & (df['job_country'] == 'United States')]
df_DA_US
```

```python
df_plot = df_DA_US['job_location'].value_counts().head(10).to_frame()

df_plot
```

```python
plt.figure(figsize = (9,6))
sns.barplot(data = df_plot, x = df_plot.index, y = 'count', palette = 'dark:b')
sns.set_theme(style = 'ticks')

plt.xticks(rotation = 45)
plt.xlabel('')
plt.ylabel('Number of Jobs')
plt.title('Counts of Job Locations for Data Analyst in the United States', fontsize = 15, fontweight = 'bold')

sns.despine()
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1,3, figsize = (12,10))

dict_column = {'job_work_from_home': 'Work from Home',
              'job_no_degree_mention': 'Job Degree Req.',
              'job_health_insurance': 'Health Insurance Offered'}

for i, (column, title) in enumerate(dict_column.items()):
    ax[i].pie(df_DA_US[column].value_counts(),
        autopct = '%.2f%%',
        startangle = 90,
        labels = ['False', 'True'],
        textprops = {'size': 'large',
                   'fontweight': 'bold'})
    ax[i].set_title(title, fontsize = 15, fontweight = 'bold')
    
plt.tight_layout()
plt.show()
```

```python
df_plot = df_DA_US['company_name'].value_counts().head(10).to_frame()

df_plot
```

```python
plt.figure(figsize = (9,6))
sns.barplot(data = df_plot, y = df_plot.index, x = 'count', palette = 'dark:b')

plt.ylabel('')
plt.xlabel('Number of Jobs')
plt.title('Counts of Companies for Data Analyst in the United States', fontsize = 14, fontweight = 'bold')

sns.despine()
plt.tight_layout()
plt.show()
```
